# DFS (Depth First Search)
1. Traverses a graph by going as deep as possible before backtracking.
2. Uses Recursion or Stack.
3. Time Complexity: O(V + E)
4. Space Complexity: O(V) (visited + recursion stack)
```
void dfs(int node) {
    visited[node] = true;

    // Process current node

    for (int next : graph[node]) {
        if (!visited[next]) {
            if (dfs(next)) return ..;
            OR
            dfs(next);
        }
    }
}

```

````python
## 1971. Find if Path Exists in Graph
from typing import List
class Solution:
    def dfs(self, source, destination, graph):
        if source == destination:
            return True
        self.visited[source] = True
        for node in graph[source]:
            if not self.visited[node]:
                if self.dfs(node, destination, graph):
                    return True
        return False
    
    def validPath(self, n, edges, source, destination):
        graph = [[] for _ in range(n)]
        for u, v  in edges:
            graph[u].append(v)
            graph[v].append(u)

        self.visited = [False]*n
        return self.dfs(source, destination, graph)
        

Sol = Solution()
edges = [[0,1],[1,2],[2,0]]
Sol.validPath(3, edges, 0, 2)
````

# BFS (Breadth First Search)

1. Traverses the graph level by level.
2. Uses a Queue.
3. Time Complexity: O(V + E)
4. Space Complexity: O(V)
```
void bfs(int start) {
    queue<int> q;

    q.push(start);
    visited[start] = true;

    while (!q.empty()) {
        int node = q.front();
        q.pop();

        // Process current node
        // cout << node << " ";

        for (int next : graph[node]) {
            if (!visited[next]) {
                visited[next] = true;
                q.push(next);
            }
        }
    }
};
```

````python
## 1971. Find if Path Exists in Graph
from typing import List
from collections import deque
class Solution:
    
    def validPath(self, n, edges, source, destination):
        graph = [[] for _ in range(n)]
        for u, v  in edges:
            graph[u].append(v)
            graph[v].append(u)

        q = deque([source])
        visited = [False]*n
        visited[source] = True
        while len(q):
            u = q.pop()
            if u == destination:
                return True
            for v in graph[u]:
                if not visited[v]:
                    visited[v] = True
                    q.append(v)
        return False

Sol = Solution()
edges = [[0,1],[1,2],[2,0]]
Sol.validPath(3, edges, 0, 2)
````
