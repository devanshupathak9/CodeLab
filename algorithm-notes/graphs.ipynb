# DFS (Depth First Search)
1. Traverses a graph by going as deep as possible before backtracking.
2. Uses Recursion or Stack.
3. Time Complexity: O(V + E)
4. Space Complexity: O(V) (visited + recursion stack)
```
void dfs(int node) {
    visited[node] = true;

    // Process current node

    for (int next : graph[node]) {
        if (!visited[next]) {
            if (dfs(next)) return ..;
            OR
            dfs(next);
        }
    }
}

```

````python
## 1971. Find if Path Exists in Graph
from typing import List
class Solution:
    def dfs(self, source, destination, graph):
        if source == destination:
            return True
        self.visited[source] = True
        for node in graph[source]:
            if not self.visited[node]:
                if self.dfs(node, destination, graph):
                    return True
        return False
    
    def validPath(self, n, edges, source, destination):
        graph = [[] for _ in range(n)]
        for u, v  in edges:
            graph[u].append(v)
            graph[v].append(u)

        self.visited = [False]*n
        return self.dfs(source, destination, graph)
        

Sol = Solution()
edges = [[0,1],[1,2],[2,0]]
Sol.validPath(3, edges, 0, 2)
````

# BFS (Breadth First Search)

1. Traverses the graph level by level.
2. Uses a Queue.
3. Time Complexity: O(V + E)
4. Space Complexity: O(V)
```
void bfs(int start) {
    queue<int> q;

    q.push(start);
    visited[start] = true;

    while (!q.empty()) {
        int node = q.front();
        q.pop();

        // Process current node
        // cout << node << " ";

        for (int next : graph[node]) {
            if (!visited[next]) {
                visited[next] = true;
                q.push(next);
            }
        }
    }
};
```

````python
## 1971. Find if Path Exists in Graph
from typing import List
from collections import deque
class Solution:
    
    def validPath(self, n, edges, source, destination):
        graph = [[] for _ in range(n)]
        for u, v  in edges:
            graph[u].append(v)
            graph[v].append(u)

        q = deque([source])
        visited = [False]*n
        visited[source] = True
        while len(q):
            u = q.pop()
            if u == destination:
                return True
            for v in graph[u]:
                if not visited[v]:
                    visited[v] = True
                    q.append(v)
        return False

Sol = Solution()
edges = [[0,1],[1,2],[2,0]]
Sol.validPath(3, edges, 0, 2)
````

# Disjoint Set (Union-Find)

A **Disjoint Set Union (DSU)** or **Union-Find** is a data structure used to maintain a collection of **non-overlapping sets**.

It supports two main operations efficiently:
- **Find** → Find the representative root of the set containing an element.
- **Union** → Merge two different sets into one.

#### Optimization
1. **Union by Rank**: Maintain the **rank (approximate height)** of each tree.

When performing a union:
- Attach the tree with the **smaller rank** under the tree with the **larger rank**.
- If both ranks are equal, attach either one and **increase the rank** of the new root by `1`.
This keeps the tree as shallow as possible.

2. **Path Compression**: During a `find(x)` operation, make every visited node directly point to the root. Using **Union by Rank + Path Compression**, each operation runs in O(α(N)).

## Template

```cpp
struct DSU {
    vector<int> parent, rank;
    DSU(int n) {
        parent.resize(n + 1);
        rank.assign(n + 1, 0);
        for (int i = 0; i <= n; i++)
            parent[i] = i;
    }
    int find(int x) {
        if (parent[x] == x)
            return x;
        // Path Compression
        return parent[x] = find(parent[x]); 
    }

    void unite(int a, int b) {
        int pa = find(a);
        int pb = find(b);
        if (pa == pb)
            return;
        // Union by Rank
        if (rank[pa] < rank[pb])
            swap(pa, pb);

        parent[pb] = pa;
        if (rank[pa] == rank[pb])
            rank[pa]++;
    }

    bool same(int a, int b) {
        return find(a) == find(b);
    }
};
```
## Complexity
| Operation | Complexity |
|-----------|------------|
| Find | O(α(N)) |
| Union | O(α(N)) |
| Space | O(N) |

> Here, **α(N)** is the Inverse Ackermann Function, which is less than 5 for all practical input sizes, so DSU operations are effectively **constant time**.

````python
# 1319. Number of Operations to Make Network Connected
class DisJointSet:
    def __init__(self, n):
        self.parent = [i for i in range(n)]
        self.rank = [0 for _ in range(n)]
        self.size = [1 for _ in range(n)]

    def find(self, x):
        if self.parent[x] == x:
            return x
        
        ### Path compression
        self.parent[x] = self.find(self.parent[x])
        return self.parent[x]
    def union(self, x, y):
        a = self.find(x)
        b = self.find(y)
        if a == b:
            return 
        if self.rank[a] < self.rank[b]:
            a,b = b,a
        self.parent[b] = a
        self.size[a] += self.size[b]

        if (self.rank[a] == self.rank[b]):
            self.rank[a] += 1

    def componentSize(self, x):
        return self.size[x]

n = 6
connections = [[0,1],[0,2],[0,3],[1,2],[1,3]]
ds = DisJointSet(n)
for u, v in connections:
    ds.union(u, v)

ds.componentSize(0)
````

# Topological Sort

A Topological Sort is an ordering of vertices in a Directed Acyclic Graph (DAG).

```
0 → 1 → 3
 \       ↑
  → 2 ───┘

**Order**: 0 1 2 3
```

### 1. DFS Method
- DFS every node.
- After processing all its neighbors, put the node into the answer.
- Reverse the answer.
```
vector<int> ans;
vector<bool> vis(n, false);

void dfs(int node) {
    vis[node] = true;

    for (int next : graph[node]) {
        if (!vis[next]) {
            dfs(next);
        }
    }

    ans.push_back(node);
}
```

### 2. **Kahn's Algorithm — BFS**
- **Use indegree**: indegree[v] = number of incoming edges
- Start with all nodes having: indegree = 0
```

for (int i = 0; i < n; i++) {
```

````python
# 210. Course Schedule II
# There are a total of numCourses courses you have to take, labeled from 0 to numCourses - 1. You are given an array prerequisites where prerequisites[i] = [ai, bi] indicates that you must take course bi first if you want to take course ai.

# For example, the pair [0, 1], indicates that to take course 0 you have to first take course 1.
# Return the ordering of courses you should take to finish all courses. If there are many valid answers, return any of them. If it is impossible to finish all courses, return an empty array.
## Kahn's Algorithm 

from collections import deque
class Solution:
    def findOrder(self, numCourses, prerequisites):
        order = []
        graph = [[] for _ in range(numCourses)]
        indegree = [0 for _ in range(numCourses)]
        for a, b in prerequisites:
            graph[b].append(a)
            indegree[a] += 1

        q = deque([])
        for i in range(len(indegree)):
            if indegree[i] == 0:
                q.append(i)

        while len(q):
            u = q.pop()
            order.append(u)

            for v in graph[u]:
                indegree[v] -= 1
                if indegree[v] == 0:
                    q.append(v)
        if len(order) != numCourses:
            return []
        return order

sol = Solution()
sol.findOrder(4, [[1,0],[2,0],[3,1],[3,2]])
````

````python
# 210. Course Schedule II
# There are a total of numCourses courses you have to take, labeled from 0 to numCourses - 1. You are given an array prerequisites where prerequisites[i] = [ai, bi] indicates that you must take course bi first if you want to take course ai.

# For example, the pair [0, 1], indicates that to take course 0 you have to first take course 1.
# Return the ordering of courses you should take to finish all courses. If there are many valid answers, return any of them. If it is impossible to finish all courses, return an empty array.

## DFS Method

from collections import deque
class Solution:
    def dfs(self, graph, node):
        self.visited[node] = True
        for v in graph[node]:
            if not self.visited[v]:
                self.dfs(graph, v)
        self.stack.append(node)

    def findOrder(self, numCourses, prerequisites):

        graph = [[] for _ in range(numCourses)]
        for a, b in prerequisites:
            graph[b].append(a)

        self.stack = []
        self.visited = [False]*numCourses
        for i in range(numCourses):
            if not self.visited[i]:
                self.dfs(graph, i)
        self.stack.reverse()
        return self.stack

sol = Solution()
sol.findOrder(4, [[1,0],[2,0],[3,1],[3,2]])
````

# Cycle Detection

#### 1. **Undirected Graph — DFS**: 

A cycle exists if we reach an already visited node that is not the parent of the current node.
```
bool dfs(int node, int parent) {
    vis[node] = true;

    for (int next : graph[node]) {
        if (!vis[next]) {
            if (dfs(next, node))
                return true;
        }
        else if (next != parent) {
            return true;
        }
    }

    return false;
}
```
---

#### 2. **Undirected Graph — BFS**: 

Same idea as DFS, but use a queue. Store (node, parent).
```
bool bfs(int start) {
    queue<pair<int, int>> q;

    q.push({start, -1});
    vis[start] = true;

    while (!q.empty()) {
        auto [node, parent] = q.front();
        q.pop();

        for (int next : graph[node]) {
            if (!vis[next]) {
                vis[next] = true;
                q.push({next, node});
            }
            else if (next != parent) {
                return true;
            }
        }
    }

    return false;
}
```
---

#### 3. **Directed Graph — DFS**

Use 3 states:
```
0 → unvisited
1 → currently in DFS path
2 → completely processed
```
```
bool dfs(int node) {
    state[node] = 1;

    for (int next : graph[node]) {
        if (state[next] == 1)
            return true;

        if (state[next] == 0 && dfs(next))
            return true;
```

#### 4. **Directed Graph — Kahn's Algorithm**

Use indegree.
- Put all indegree == 0 nodes in queue.
- Remove them and decrease neighbors' indegree.
- If fewer than n nodes are processed → cycle.
