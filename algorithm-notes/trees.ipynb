# Binary Tree
## Binary Tree Level Order Traversal

```python
from collections import deque

class TreeNode:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

class Solution:
    def levelOrder(self, root):
        result = []
        q = deque([root])
        while len(q):
            size = len(q)
            level_nodes = []
            while size > 0:
                curr = q.popleft()
                size -= 1
                if not curr: 
                    continue
                level_nodes.append(curr.val)
                if curr.left != None:
                    q.append(curr.left)
                if curr.right != None: 
                    q.append(curr.right)
            if level_nodes:
                result.append(level_nodes)
        return result

root = TreeNode(6)
root.left = TreeNode(12)
root.right = TreeNode(5)
root.left.left = TreeNode(9)
root.right.left = TreeNode(15)
root.left.right = TreeNode(32)
root.right.right = TreeNode(87)

sol = Solution()
print(sol.levelOrder(root))
```

## Maximum Depth of Binary Tree
A binary tree's maximum depth is the number of nodes along the longest path from the root node down to the farthest leaf node.

```python
class TreeNode:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

class Solution:
    def maxDepth(self, root):
        if root == None:
            return 0
        return 1 + max(self.maxDepth(root.left), self.maxDepth(root.right))
    
root = TreeNode(5)
root.left = TreeNode(12)
root.right = TreeNode(55)
root.left.left = TreeNode(9)
root.right.left = TreeNode(15)
root.left.right = TreeNode(32)
root.right.right = TreeNode(87)
root.left.right.left = TreeNode(32)
root.right.right.right = TreeNode(87)

sol = Solution()
sol.maxDepth(root)
```
