## 3. Important Bit Tricks

### Check if Power of 2
```python
def is_power_of_two(n):
    return n > 0 and (n & (n - 1)) == 0
```

### Count Set Bits (Brian Kernighan's Algorithm)
```python
def count_set_bits(n):
    count = 0
    while n:
        n &= (n - 1)  # Clears lowest set bit
        count += 1
    return count

# Alternative: bin(n).count('1')
```

### Toggle, Set, Clear, Check Bit
```python
# Set bit at position i (0-indexed)
n | (1 << i)

# Clear bit at position i
n & ~(1 << i)

# Toggle bit at position i
n ^ (1 << i)

# Check if bit i is set
(n & (1 << i)) != 0
```

### Find Position of Rightmost Set Bit
```python
# Position (1-indexed)
pos = (n & -n).bit_length()  # or
pos = int(math.log2(n & -n)) + 1
```

## 4. Common Bit Hacks

| Operation | Formula |
|-----------|---------|
| Isolate lowest set bit | `n & -n` |
| Clear lowest set bit | `n & (n-1)` |
| Clear all bits from LSB to i | `n & ~((1 << (i+1)) - 1)` |
| Clear all bits from i to MSB | `n & ((1 << i) - 1)` |
| Set bits in range i to j | `((1 << (j-i+1)) - 1) << i` |

## 6. Important Patterns

### Find Missing Number (1 to n)
```python
def missing_number(nums):
    n = len(nums)
    xor_all = 0
    for i in range(n+1):
        xor_all ^= i
    for num in nums:
        xor_all ^= num
    return xor_all
```

### Find Two Non-Repeating Numbers
```python
def find_two_unique(nums):
    xor_all = 0
    for num in nums:
        xor_all ^= num
    
    # Get rightmost set bit
    rightmost = xor_all & -xor_all
    
    x, y = 0, 0
    for num in nums:
        if num & rightmost:
            x ^= num
        else:
            y ^= num
    return x, y
```

## 7. Bit Masking Operations
```python
# Create mask of n ones
mask = (1 << n) - 1

# Extract bits i to j (j > i)
value = (num >> i) & ((1 << (j-i+1)) - 1)

# Replace bits i to j with m
num = num & ~(((1 << (j-i+1)) - 1) << i) | (m << i)
```

````python
# Single Number
# You are given a non-empty array of integers nums. Every integer appears twice except for one.
# Return the integer that appears only once.
# You must implement a solution with O(n) runtime complexity and use only O(1) extra space.

def singleNumber(nums):
    x = 0
    for num in nums:
        x = x^num
    return x

nums = [7,6,6,7,8]
singleNumber(nums)
````

````python
def missing_number(nums):
    n = len(nums)
    xor_all = 0
    for i in range(1, n+2):
        xor_all ^= i
    for num in nums:
        xor_all ^= num
    return xor_all

nums = [1,2,3,5]
missing_number(nums)
````

````python
# Number of 1 Bits
# You are given an unsigned integer n. Return the number of 1 bits in its binary representation.
# You may assume n is a non-negative integer which fits within 32-bits.

def hammingWeight(n):
    count = 0
    for i in range(32):
        if (1<<i&n):
            count += 1
    return count

def hammingWeight2(n):
    count = 0
    for i in range(32):
        if (n>>i&1):
            count += 1
    return count

n = 128
hammingWeight(n), hammingWeight2(n)
````
