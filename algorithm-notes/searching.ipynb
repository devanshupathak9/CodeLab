# Binary Search

Binary Search works on a sorted search space by repeatedly dividing it into two halves.
At each step, eliminate the half that cannot contain the answer → O(log n).
```
int binarySearch(vector<int>& nums, int target) {
    int l = 0, r = nums.size() - 1;

    while (l <= r) {
        int mid = l + (r - l) / 2;

        if (nums[mid] == target)
            return mid;
        else if (nums[mid] < target)
            l = mid + 1;
        else
            r = mid - 1;
    }

    return -1;
}
```

````python
def binary_search(arr, target):
    i = 0
    j = len(arr) - 1
    while i <= j:
        mid = i + (j-i)//2
        if arr[mid] == target:
            return mid
        if arr[mid] > target:
            j = mid - 1
        else:
            i = mid + 1

arr = [1, 3, 5, 7, 9, 14, 35, 68, 89]
binary_search(arr, 7)
````

### Left and Right Occurance

````python
# 34. Find First and Last Position of Element in Sorted Array
# Given an array of integers nums sorted in non-decreasing order, find the starting and ending position of a given target value.
# If target is not found in the array, return [-1, -1].
# You must write an algorithm with O(log n) runtime complexity.

def left_occurance(arr, target):
    i = 0
    j = len(arr) - 1
    idx = None
    while i <= j:
        mid = i + (j - i)//2
        if arr[mid] >= target:
            idx = mid
            j = mid - 1
        else: 
            i = mid + 1
    return idx

def right_occurance(arr, target):
    i = 0
    j = len(arr) - 1
    idx = None
    while i <= j:
        mid = i + (j - i)//2
        if arr[mid] <= target:
            idx = mid
            i = mid + 1
        else: 
            j = mid - 1
    return idx

arr = [1,1,1,3,3,5,5,7,7,8,8,10]
target = 8
left_occurance(arr, target), right_occurance(arr, target)
````
