# Backtracking

```
def backtrack(path, choices, result):
    if is_valid_solution(path):
        result.append(path[:]) 
        return

    for i in range(len(choices)):
        if not is_safe(choices[i], path):
            continue

        path.append(choices[i])
        backtrack(path, choices, result) 
        path.pop()
```


### Leetcode 78. Subsets

Given an integer array nums of unique elements, return all possible subsets (the power set).

The solution set must not contain duplicate subsets. Return the solution in any order.

````python
class Solution:
    def __init__(self):
        self.answer = []
        
    def subsets(self, nums):
        def generate_subsets(start, array):
            self.answer.append(array.copy())
            for i in range(start, len(nums)):
                array.append(nums[i])
                generate_subsets(i + 1, array)
                array.pop()
        generate_subsets(0, [])
        return self.answer

sol = Solution()
nums = [1, 2, 3, 4]
sol.subsets(nums)
````

### Leetcode 90. Subsets II

Given an integer array nums that may contain duplicates, return all possible subsets (the power set).

The solution set must not contain duplicate subsets. Return the solution in any order.

````python
class Solution:
    def __init__(self):
        self.answer = []
        self.hash = {}
        
    def subsetsWithDup(self, nums):
        nums.sort()
        def generate_subsets(start, array):
            hashvalue = ""
            for val in array:
                hashvalue += f"{val} "
            if not self.hash.get(hashvalue):
                self.hash[hashvalue] = True
                self.answer.append(array.copy())
            for i in range(start, len(nums)):
                array.append(nums[i])
                generate_subsets(i + 1, array)
                array.pop()
        generate_subsets(0, [])
        return self.answer

    def subsetsWithDup2(self, nums):
        nums.sort()
        def generate_subsets(start, array):
            self.answer.append(array.copy())
            for i in range(start, len(nums)):
                if i > start and nums[i] == nums[i-1]:
                    continue
                array.append(nums[i])
                generate_subsets(i + 1, array)
                array.pop()
        generate_subsets(0, [])
        return self.answer
    

sol = Solution()
nums = [1, 2, 2, 4]
# print(sol.subsetsWithDup(nums))
print(sol.subsetsWithDup2(nums))
````
